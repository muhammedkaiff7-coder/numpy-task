```python
import pandas as pd

data = {
    "Name": [
        " Rahul ",
        "PRIYA",
        "arjun",
        "Sneha",
        "rahul ",
        "Akhil",
        None,
        " Meera",
        "PRIYA"
    ],

    "Age": [
        "22",
        "23",
        None,
        "21",
        "22",
        "24",
        "25",
        "23",
        "23"
    ],

    "City": [
        " kochi",
        "Kochi ",
        "KOCHI",
        "Trivandrum",
        None,
        " kochi ",
        "Calicut",
        "CALICUT ",
        "Kochi "
    ],

    "Course": [
        "Data Science",
        "data science",
        "DATA SCIENCE ",
        "Python",
        None,
        "Python ",
        "Data Science",
        "python",
        "data science"
    ],

    "Salary": [
        "25000",
        "30000",
        "28000",
        None,
        "25000",
        "32000",
        "35000",
        "30000",
        "30000"
    ],

    "Join_Date": [
        "2025-01-10",
        "15/01/2025",
        "2025-02-05",
        None,
        "2025-01-10",
        "20/02/2025",
        "2025-03-01",
        "invalid",
        "15/01/2025"
    ]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)



#  INSPECT THE DATA


print("\nData Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())



# 3. STRING CLEANING

# Remove leading/trailing spaces
df["Name"] = df["Name"].str.strip()
df["City"] = df["City"].str.strip()
df["Course"] = df["Course"].str.strip()

# Convert to consistent format
df["Name"] = df["Name"].str.title()
df["City"] = df["City"].str.title()
df["Course"] = df["Course"].str.title()

print("\nAfter String Cleaning:")
print(df)



#  MISSING VALUE HANDLING
 

# Fill Age using median
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill Salary using median
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Fill City
df["City"] = df["City"].fillna("Unknown")

# Fill Course
df["Course"] = df["Course"].fillna("Not Specified")


df["Name"] = df["Name"].fillna("Unknown")

print("\nAfter Handling Missing Values:")
print(df)



# 5. DUPLICATE REMOVAL


print("\nDuplicate Rows:")
print(df[df.duplicated()])

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()



# 6. DATA TYPE CONVERSION


df["Age"] = df["Age"].astype(int)
df["Salary"] = df["Salary"].astype(float)



#  DATE CONVERSION
 

# Invalid date becomes NaT instead of giving an error
df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("\nAfter Date Conversion:")
print(df)


# . VERIFY MISSING VALUES


print("\nMissing Values After Cleaning:")
print(df.isnull().sum())



# VERIFY DUPLICATES


print("\nNumber of Duplicate Rows After Cleaning:")
print(df.duplicated().sum())



#  FINAL CLEANED DATAFRAME


print("\nFinal Cleaned DataFrame:")
print(df)
```
