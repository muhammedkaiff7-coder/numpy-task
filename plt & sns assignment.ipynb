```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Sunaj\Desktop\New folder (3)\pythonnn\sales visualization.csv")
df
```

1.SALES BY CATEGORY

```python
category_sales = df.groupby("Category")["Sales"].sum()
```

```python
plt.figure(figsize=(8,5))
plt.bar(category_sales.index, category_sales.values)
plt.title("Total sales by category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

2.SALES BY REGION

```python
region_sales = df.groupby("Region") ["Sales"].sum()
```

```python
plt.figure(figsize=(8,5))
plt.bar(region_sales.index,region_sales.values)
plt.title("Total Sales byRegion")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
3.SALES BY CUSTOMER SEGMENT
```

```python
segment_sales = df.groupby("Customer_Segment")["Sales"].sum()
```

```python
plt.figure(figsize=(6,6))
plt.pie(segment_sales.values,labels=segment_sales.index,autopct="%1.1f%%")
plt.title("Sales by Customer Segment")
plt.show()
```

4.MONTHLY SALES TREND

```python
monthly_sales=df.groupby("Month")["Sales"].sum()
```

```python
month_order=["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
montly_sales = monthly_sales.reindex(month_order)
plt.figure(figsize=(10,5))
plt.plot(monthly_sales.index,monthly_sales.values,marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid()
plt.show()
```

6.  TOP 5 PRODUCTS BY SALES

```python
product_sales=df.groupby("Product")["Sales"].sum()
```

```python
top_5_products = product_sales.sort_values(ascending=False).head(5)
```

```python
plt.figure(figsize=(6,4))
plt.barh(top_5_products.index,top_5_products.values)
plt.title("top 5 products by sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.gca().invert_yaxis()
plt.show()
```

7.SALES BY CITY

```python
city_sales = df.groupby("City")["Sales"].sum()
```

```python
plt.bar(city_sales.index,city_sales.values)
plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

8.QUANTITY VS SALES

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Quantity"],df["Sales"])
plt.title("Quantity vs Sales")
plt.xlabel("Quantity Sold")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

9.SALES BY CATEGORY BY REGION

```python
sales_pivot = df.pivot_table(values="Sales",index="Category",columns="Region",aggfunc="sum")
plt.figure(figsize=(10,6))
plt.imshow(sales_pivot, aspect="auto")
plt.colorbar(label="Total Sales")
plt.xticks
plt.yticks
plt.title("Sales by Category and Region")
plt.xlabel("Region")
plt.ylabel("Category")
plt.show()
```

1.SALS DISTRIBUTION

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Sales"],bins=10)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```
